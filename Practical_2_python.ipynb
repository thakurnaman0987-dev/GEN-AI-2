```python
# 1. Create an empty list
shopping_list = []

# 2. Store items in the shopping list
shopping_list = ["Watch", "Laptop", "Shoes", "Pen", "Clothes"]

# 3. Add Football to the shopping list
shopping_list.append("Football")

# 4. Print the last item
print("Last item:", shopping_list[-1])

# 5. Print the entire shopping list
print("Shopping List:", shopping_list)

# 6. Print important items: Laptop and Shoes
print("Important items:", shopping_list[1], "and", shopping_list[2])

# 7. Change Pen to Notebook
shopping_list[3] = "Notebook"

print("After changing Pen to Notebook:")
print(shopping_list)

# 8. Delete the least important item
# Here, we consider Watch as the least important item
shopping_list.remove("Watch")

print("Final Shopping List:")
print(shopping_list)
```
