```python
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __repr__(self):
        return f"Point({self.x}, {self.y})"


# Creating object
p1 = Point(2, 3)

# Printing object
print(p1)
```

```python
class Animal:

    def __init__(self, name):
        self.name = name

    def speak(self):
        print(self.name)


class Dog(Animal):
    pass


# Creating object
d1 = Dog("Rex")

# Calling inherited method
d1.speak()
```

```python
class Cat:

    def sound(self):
        print("Meow")


class Fox:

    def sound(self):
        print("Wa-pa-pa-pa-pa-pow!")


# Creating objects
c1 = Cat()
f1 = Fox()

# Calling sound methods
c1.sound()
f1.sound()
```

```python
class ScoreBoard:

    def __init__(self, score):
        self.__score = score

    def get_score(self):
        return self.__score


# Creating object
s1 = ScoreBoard(0)

# Printing score
print(s1.get_score())
```
